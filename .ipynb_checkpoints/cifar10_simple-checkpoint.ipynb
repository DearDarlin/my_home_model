# CIFAR-10: простая CNN на PyTorch

```python
import random
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

torch.manual_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

CLASSES = ["Airplane", "Automobile", "Bird", "Cat", "Deer", "Dog", "Frog", "Horse", "Ship", "Truck"]
```

## 1. Данные
Аугментация (только для train): случайный сдвиг (`RandomCrop`) и отражение (`RandomHorizontalFlip`).
`Normalize` приводит значения пикселей к среднему ~0 и std ~1 — так сеть обучается быстрее.

```python
MEAN, STD = (0.4914, 0.4822, 0.4465), (0.2470, 0.2435, 0.2616)

train_tf = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])
test_tf = transforms.Compose([transforms.ToTensor(), transforms.Normalize(MEAN, STD)])

train_data = datasets.CIFAR10("./data", train=True, download=True, transform=train_tf)
test_data = datasets.CIFAR10("./data", train=False, download=True, transform=test_tf)

train_loader = DataLoader(train_data, batch_size=128, shuffle=True)
test_loader = DataLoader(test_data, batch_size=256)
```

## 2. Модель
Три одинаковых блока `Conv → BatchNorm → ReLU → MaxPool` (фильтров 64 → 128 → 256, размер 32 → 16 → 8 → 4),
затем `Flatten → Dropout → Linear(4096 → 10)`.

```python
def block(c_in, c_out):
    return nn.Sequential(
        nn.Conv2d(c_in, c_out, kernel_size=3, padding=1),
        nn.BatchNorm2d(c_out),
        nn.ReLU(),
        nn.MaxPool2d(2),
    )

model = nn.Sequential(
    block(3, 64),
    block(64, 128),
    block(128, 256),
    nn.Flatten(),
    nn.Dropout(0.3),
    nn.Linear(256 * 4 * 4, 10),   # 10 логитов, Softmax не нужен (он внутри CrossEntropyLoss)
).to(device)

print(model)
print("Параметров:", sum(p.numel() for p in model.parameters()))
```

## 3. Обучение
Loss — `CrossEntropyLoss`, оптимизатор — Adam (lr=0.001), 20 эпох.
`StepLR` уменьшает learning rate в 10 раз на 15-й эпохе — модель «доводится» мелкими шагами.

```python
EPOCHS = 20
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=15, gamma=0.1)

def accuracy(loader):
    model.eval()
    correct = 0
    with torch.no_grad():
        for x, y in loader:
            correct += (model(x.to(device)).argmax(1).cpu() == y).sum().item()
    return correct / len(loader.dataset)

for epoch in range(1, EPOCHS + 1):
    model.train()
    total_loss = 0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        optimizer.zero_grad()
        loss = criterion(model(x), y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * x.size(0)
    scheduler.step()
    print(f"Epoch {epoch:02d}/{EPOCHS} | train loss={total_loss/len(train_data):.4f} | test acc={accuracy(test_loader):.4f}")
```

## 4. Test Accuracy

```python
test_acc = accuracy(test_loader)
print(f"Test accuracy: {test_acc*100:.2f}%")
print("Бонус >80% выполнен ✅" if test_acc > 0.80 else "Бонус >80% не выполнен ❌")
```

```python
# Предсказания и уверенность для всех тестовых картинок
model.eval()
probs, labels = [], []
with torch.no_grad():
    for x, y in test_loader:
        probs.append(torch.softmax(model(x.to(device)), dim=1).cpu())
        labels.append(y)
probs, labels = torch.cat(probs), torch.cat(labels)
confs, preds = probs.max(dim=1)

def show(img):  # обратная нормализация для отображения
    mean, std = torch.tensor(MEAN).view(3, 1, 1), torch.tensor(STD).view(3, 1, 1)
    return (img * std + mean).clamp(0, 1).permute(1, 2, 0).numpy()
```

## 5. 20 случайных изображений

```python
idxs = random.sample(range(len(test_data)), 20)
plt.figure(figsize=(15, 12))
for n, i in enumerate(idxs):
    real, pred, conf = CLASSES[labels[i]], CLASSES[preds[i]], confs[i].item() * 100
    ok = real == pred
    print(f"REAL: {real} | AI: {pred} {'✅' if ok else '❌'} | Confidence: {conf:.0f}%")
    plt.subplot(4, 5, n + 1)
    plt.imshow(show(test_data[i][0]))
    plt.title(f"REAL: {real}\nAI: {pred} {'✓' if ok else '✗'} | {conf:.0f}%", color="green" if ok else "red", fontsize=10)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

## 6. 5 самых уверенных ошибок

```python
wrong = (preds != labels).nonzero(as_tuple=True)[0]
top5 = wrong[confs[wrong].argsort(descending=True)[:5]]

plt.figure(figsize=(16, 4))
for n, i in enumerate(top5):
    real, pred, conf = CLASSES[labels[i]], CLASSES[preds[i]], confs[i].item() * 100
    print(f"REAL: {real} | AI: {pred} ❌ | Confidence: {conf:.1f}%")
    plt.subplot(1, 5, n + 1)
    plt.imshow(show(test_data[i.item()][0]))
    plt.title(f"REAL: {real}\nAI: {pred} | {conf:.1f}%", color="red", fontsize=10)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### Почему сеть ошиблась? (впишите свой разбор после запуска)
Возможные причины:
- картинки всего 32×32 — мелкие детали (морда, колёса, крылья) почти не видны;
- похожие силуэты и позы (кошка/собака, олень/лошадь);
- фон подсказывает неверный класс (синий фон → самолёт/корабль, трава → олень/лошадь);
- нетипичный ракурс или объект обрезан;
- сомнительная разметка — иногда даже человек не уверен;
- Softmax может давать высокую уверенность и при неправильном ответе.

## 7. Confusion Matrix

```python
cm = torch.zeros(10, 10, dtype=torch.long)
for t, p in zip(labels, preds):
    cm[t, p] += 1

plt.figure(figsize=(8, 7))
plt.imshow(cm, cmap="Blues")
plt.colorbar()
plt.xticks(range(10), CLASSES, rotation=45, ha="right")
plt.yticks(range(10), CLASSES)
plt.xlabel("Предсказано")
plt.ylabel("Истина")
plt.title("Confusion Matrix")
for i in range(10):
    for j in range(10):
        plt.text(j, i, cm[i, j].item(), ha="center", va="center",
                 color="white" if cm[i, j] > cm.max() / 2 else "black", fontsize=8)
plt.tight_layout()
plt.show()

# Самые частые путаницы
pairs = [(cm[i, j].item(), CLASSES[i], CLASSES[j]) for i in range(10) for j in range(10) if i != j]
print("Топ-5 путаниц (истина → предсказание):")
for n, a, b in sorted(pairs, reverse=True)[:5]:
    print(f"  {a} → {b}: {n}")
```

### Какие классы путаются и почему (сверьте со своей матрицей)
- **Cat ↔ Dog** — самая частая пара: оба с шерстью, четырьмя лапами, похожие позы и фон; на 32×32 морда — несколько пикселей.
- **Deer ↔ Horse** — похожее вытянутое тело на тонких ногах, коричневый цвет, трава на фоне.
- **Automobile ↔ Truck** — оба на колёсах и на дороге, отличаются только размером/формой кузова.
- **Airplane ↔ Ship / Bird** — синий фон неба или воды, объект маленький.

## 8. Зачем каждый слой и приём
- **Conv2d (3×3, padding=1)** — обучаемые фильтры, ищут локальные паттерны (границы → текстуры → части объектов). Веса общие для всей картинки, поэтому параметров намного меньше, чем у Linear. `padding=1` сохраняет размер карты.
- **Рост фильтров 64 → 128 → 256** — чем глубже слой, тем больше разных абстрактных признаков нужно.
- **BatchNorm2d** — нормализует активации по батчу: обучение стабильнее и быстрее.
- **ReLU** — нелинейность; без неё стопка свёрток была бы одной линейной операцией.
- **MaxPool2d(2)** — уменьшает карту в 2 раза: меньше вычислений, небольшая устойчивость к сдвигам, больше «поле зрения» у следующих слоёв.
- **Flatten** — превращает карты признаков в вектор для Linear.
- **Dropout(0.3)** — случайно выключает часть нейронов при обучении, борьба с переобучением.
- **Linear(4096 → 10)** — 10 логитов, по одному на класс. Softmax не ставим: он уже внутри `CrossEntropyLoss`; вручную применяем его только для показа уверенности.
- **Data Augmentation** — сдвиги и отражения «увеличивают» датасет и снижают переобучение.
- **Normalize** — входные данные с нулевым средним и единичным std.
- **Adam + StepLR** — Adam быстро сходится; уменьшение lr в конце помогает точнее дообучиться.
- **model.eval() / torch.no_grad()** — на тесте отключаем Dropout, BatchNorm использует накопленную статистику, градиенты не считаем.
