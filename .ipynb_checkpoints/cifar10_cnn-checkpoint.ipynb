# CIFAR-10: собственная CNN на PyTorch

**План:** данные + аугментация → VGG-подобная CNN с BatchNorm и Dropout → обучение (SGD + OneCycleLR) →
test accuracy → 20 случайных предсказаний → 5 самых уверенных ошибок → confusion matrix → выводы.

Готовые архитектуры и предобученные веса **не используются**: всё собрано из `Conv2d`, `BatchNorm2d`, `ReLU`, `MaxPool2d`, `Dropout`, `Linear`.

```python
import random
import time
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms

SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu")
print("Device:", device)

CLASSES = ["Airplane", "Automobile", "Bird", "Cat", "Deer", "Dog", "Frog", "Horse", "Ship", "Truck"]
```

## 1. Данные и Data Augmentation

- **Normalize** по средним/std каналов CIFAR-10 — входы с нулевым средним и единичной дисперсией, обучение стабильнее.
- **RandomCrop(32, padding=4)** — случайный сдвиг объекта: сеть учится не привязываться к точной позиции.
- **RandomHorizontalFlip** — лошадь/кошка, смотрящая влево или вправо, остаётся тем же классом.
- Аугментация только на train. Val и test — без неё, иначе оценка будет искажена.
- Из train (50 000) выделяем 5 000 на validation: по ним выбираем лучшую эпоху, **test трогаем один раз в конце**.

```python
MEAN = (0.4914, 0.4822, 0.4465)
STD = (0.2470, 0.2435, 0.2616)

train_tf = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])
eval_tf = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

train_aug = datasets.CIFAR10("./data", train=True, download=True, transform=train_tf)
train_plain = datasets.CIFAR10("./data", train=True, download=True, transform=eval_tf)
test_data = datasets.CIFAR10("./data", train=False, download=True, transform=eval_tf)

g = torch.Generator().manual_seed(SEED)
perm = torch.randperm(len(train_aug), generator=g).tolist()
val_idx, train_idx = perm[:5000], perm[5000:]

train_data = Subset(train_aug, train_idx)     # с аугментацией
val_data = Subset(train_plain, val_idx)       # без аугментации

BATCH_SIZE = 128
train_loader = DataLoader(train_data, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True)
val_loader = DataLoader(val_data, batch_size=256, num_workers=2, pin_memory=True)
test_loader = DataLoader(test_data, batch_size=256, num_workers=2, pin_memory=True)
print(len(train_data), len(val_data), len(test_data))
```

## 2. Архитектура

Три блока по схеме `Conv-BN-ReLU → Conv-BN-ReLU → MaxPool → Dropout`, число фильтров растёт 64 → 128 → 256, размер карты падает 32 → 16 → 8 → 4.
Затем классификатор `Flatten → Dropout → Linear(4096→256) → ReLU → Dropout → Linear(256→10)`.

Зачем каждый компонент — в разделе «Объяснение слоёв» в конце.

```python
def conv_block(c_in, c_out, p_drop):
    return nn.Sequential(
        nn.Conv2d(c_in, c_out, kernel_size=3, padding=1, bias=False),  # bias не нужен: BN его всё равно вычтет
        nn.BatchNorm2d(c_out),
        nn.ReLU(inplace=True),
        nn.Conv2d(c_out, c_out, kernel_size=3, padding=1, bias=False),
        nn.BatchNorm2d(c_out),
        nn.ReLU(inplace=True),
        nn.MaxPool2d(2),
        nn.Dropout(p_drop),
    )

class CifarCNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.features = nn.Sequential(
            conv_block(3, 64, 0.10),     # 32x32 -> 16x16
            conv_block(64, 128, 0.20),   # 16x16 -> 8x8
            conv_block(128, 256, 0.30),  # 8x8   -> 4x4
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.4),
            nn.Linear(256 * 4 * 4, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.4),
            nn.Linear(256, num_classes),   # логиты; Softmax не ставим — его включает CrossEntropyLoss
        )

    def forward(self, x):
        return self.classifier(self.features(x))

model = CifarCNN().to(device)
print(model)
print(f"Обучаемых параметров: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")
print("Проверка формы:", model(torch.rand(2, 3, 32, 32, device=device)).shape)
```

## 3. Обучение

- **Loss:** `CrossEntropyLoss` (мультиклассовая классификация, на вход — логиты).
- **Optimizer:** SGD + momentum 0.9 + Nesterov, `weight_decay=5e-4` (L2-регуляризация против переобучения).
- **Scheduler:** `OneCycleLR` — lr быстро растёт до максимума, затем плавно падает почти до нуля. Это даёт быструю сходимость и точную «доводку» в конце.
- Сохраняем веса лучшей по validation эпохи.

```python
EPOCHS = 30
MAX_LR = 0.1

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=MAX_LR, momentum=0.9, nesterov=True, weight_decay=5e-4)
scheduler = torch.optim.lr_scheduler.OneCycleLR(
    optimizer, max_lr=MAX_LR, epochs=EPOCHS, steps_per_epoch=len(train_loader), pct_start=0.2)

def train_one_epoch(model, loader):
    model.train()
    total_loss = correct = total = 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        optimizer.zero_grad()
        logits = model(x)
        loss = criterion(logits, y)
        loss.backward()
        optimizer.step()
        scheduler.step()
        total_loss += loss.item() * x.size(0)
        correct += (logits.argmax(1) == y).sum().item()
        total += y.size(0)
    return total_loss / total, correct / total

@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    total_loss = correct = total = 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        logits = model(x)
        total_loss += criterion(logits, y).item() * x.size(0)
        correct += (logits.argmax(1) == y).sum().item()
        total += y.size(0)
    return total_loss / total, correct / total
```

```python
history = {"train_loss": [], "train_acc": [], "val_loss": [], "val_acc": []}
best_val, best_state = 0.0, None

for epoch in range(1, EPOCHS + 1):
    t0 = time.time()
    tl, ta = train_one_epoch(model, train_loader)
    vl, va = evaluate(model, val_loader)
    for k, v in zip(history, (tl, ta, vl, va)):
        history[k].append(v)
    if va > best_val:
        best_val = va
        best_state = {k: v.detach().clone() for k, v in model.state_dict().items()}
    print(f"Epoch {epoch:02d}/{EPOCHS} | train loss={tl:.4f} acc={ta:.4f} | val loss={vl:.4f} acc={va:.4f} | {time.time()-t0:.0f}s")

model.load_state_dict(best_state)
print(f"\nЛучшая val accuracy: {best_val:.4f}")
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(history["train_loss"], label="train"); ax[0].plot(history["val_loss"], label="val")
ax[0].set_title("Loss"); ax[0].set_xlabel("epoch"); ax[0].legend()
ax[1].plot(history["train_acc"], label="train"); ax[1].plot(history["val_acc"], label="val")
ax[1].set_title("Accuracy"); ax[1].set_xlabel("epoch"); ax[1].legend()
plt.tight_layout(); plt.show()
```

## 4. Финальная оценка на TEST

```python
test_loss, test_acc = evaluate(model, test_loader)
print(f"Test loss: {test_loss:.4f}")
print(f"Test accuracy: {test_acc:.4f}  ({test_acc*100:.2f}%)")
print("Бонус (>80%) выполнен ✅" if test_acc > 0.80 else "Бонус (>80%) пока не выполнен ❌")
```

```python
# Предсказания и уверенность для всей тестовой выборки (понадобятся дальше)
model.eval()
all_probs, all_labels = [], []
with torch.no_grad():
    for x, y in test_loader:
        all_probs.append(F.softmax(model(x.to(device)), dim=1).cpu())
        all_labels.append(y)
probs = torch.cat(all_probs)
labels = torch.cat(all_labels)
confs, preds = probs.max(dim=1)

mean_t = torch.tensor(MEAN).view(3, 1, 1)
std_t = torch.tensor(STD).view(3, 1, 1)

def show_img(t):
    """Денормализация тензора [3,32,32] -> картинка для imshow."""
    return (t * std_t + mean_t).clamp(0, 1).permute(1, 2, 0).numpy()
```

## 5. 20 случайных изображений

```python
rng = random.Random()  # без фиксированного seed — при каждом запуске новые картинки
idxs = rng.sample(range(len(test_data)), 20)

fig, axes = plt.subplots(4, 5, figsize=(15, 13))
for ax, i in zip(axes.flat, idxs):
    img, _ = test_data[i]
    real, pred, conf = CLASSES[labels[i]], CLASSES[preds[i]], confs[i].item() * 100
    ok = real == pred
    line = f"REAL: {real} | AI: {pred} {'✅' if ok else '❌'} | Confidence: {conf:.0f}%"
    print(line)
    ax.imshow(show_img(img))
    ax.set_title(f"REAL: {real}\nAI: {pred} {'✓' if ok else '✗'} | {conf:.0f}%",
                 color="green" if ok else "red", fontsize=10)
    ax.axis("off")
plt.tight_layout(); plt.show()
```

## 6. Топ-5 ошибок с максимальной уверенностью

```python
wrong = (preds != labels).nonzero(as_tuple=True)[0]
top5 = wrong[confs[wrong].argsort(descending=True)[:5]]

fig, axes = plt.subplots(1, 5, figsize=(18, 4.5))
for ax, i in zip(axes, top5):
    img, _ = test_data[i.item()]
    real, pred, conf = CLASSES[labels[i]], CLASSES[preds[i]], confs[i].item() * 100
    print(f"#{i.item():5d} REAL: {real} | AI: {pred} ❌ | Confidence: {conf:.1f}%")
    ax.imshow(show_img(img), interpolation="nearest")
    ax.set_title(f"REAL: {real}\nAI: {pred} ❌ | {conf:.1f}%", color="red", fontsize=11)
    ax.axis("off")
plt.tight_layout(); plt.show()
```

### Почему сеть ошиблась? (заполните после просмотра своих картинок)

Картинки у каждого получатся свои (зависит от обучения), поэтому ниже — список типичных причин. Для каждой из 5 картинок выберите подходящую и напишите 1–2 предложения.

1. **Низкое разрешение 32×32.** Мелкие детали (морда, уши, колёса, крылья) сливаются в несколько пикселей — кошка и собака, олень и лошадь выглядят почти одинаково.
2. **Похожий силуэт и поза.** Четвероногое животное на травянистом фоне: у кошки/собаки/оленя/лошади схожая форма тела.
3. **Фон как «подсказка».** Сеть могла выучить корреляции: синий фон → самолёт или корабль, трава → олень/лошадь/лягушка. Если фон «не тот», она ошибается.
4. **Нетипичный ракурс, обрезанный объект, часть объекта в кадре.**
5. **Неоднозначная или некорректная разметка** — иногда даже человеку трудно сказать, какой тут класс (или метка в датасете сомнительна).
6. **Переуверенность.** Softmax выдаёт высокую вероятность, даже когда сеть не права: уверенность ≠ правильность, сеть плохо откалибрована (особенно без label smoothing).

## 7. Confusion Matrix

```python
num_classes = len(CLASSES)
cm = torch.zeros(num_classes, num_classes, dtype=torch.long)
for t, p in zip(labels, preds):
    cm[t, p] += 1
cm_np = cm.numpy()
cm_norm = cm_np / cm_np.sum(axis=1, keepdims=True)

fig, ax = plt.subplots(1, 2, figsize=(18, 7))
for a, data, title, fmt in [(ax[0], cm_np, "Confusion Matrix (количество)", "d"),
                            (ax[1], cm_norm, "Confusion Matrix (доля от истинного класса)", ".2f")]:
    im = a.imshow(data, cmap="Blues")
    a.set_xticks(range(num_classes)); a.set_yticks(range(num_classes))
    a.set_xticklabels(CLASSES, rotation=45, ha="right"); a.set_yticklabels(CLASSES)
    a.set_xlabel("Предсказано"); a.set_ylabel("Истина"); a.set_title(title)
    thr = data.max() / 2
    for i in range(num_classes):
        for j in range(num_classes):
            a.text(j, i, format(data[i, j], fmt), ha="center", va="center",
                   color="white" if data[i, j] > thr else "black", fontsize=8)
    plt.colorbar(im, ax=a, fraction=0.046)
plt.tight_layout(); plt.show()
```

```python
# Самые частые ошибки и точность по классам
pairs = [(cm_np[i, j], CLASSES[i], CLASSES[j]) for i in range(num_classes) for j in range(num_classes) if i != j]
print("Топ-8 путаниц (истина → предсказание):")
for n, a, b in sorted(pairs, reverse=True)[:8]:
    print(f"  {a:>10} → {b:<10}: {n}")

print("\nТочность по классам:")
for i, c in sorted(enumerate(CLASSES), key=lambda t: cm_norm[t[0], t[0]]):
    print(f"  {c:>10}: {cm_norm[i, i]*100:.1f}%")
```

### Какие классы путаются и почему

Типичная картина для CIFAR-10 (сверьте со своей матрицей):

- **Cat ↔ Dog** — самая частая пара. Оба: четыре лапы, шерсть, похожие позы и фон (дом, трава); при 32×32 морда — это несколько пикселей. Cat обычно и самый слабый класс.
- **Deer ↔ Horse** — вытянутое тело на ножках, коричневая окраска, фон трава/лес.
- **Automobile ↔ Truck** — оба колёсные транспортные средства на дороге, различие только в размере/форме кузова, а масштаб на картинке неочевиден.
- **Airplane ↔ Ship (и Bird)** — синий фон неба/воды; сеть может опираться на цвет фона, а не на сам объект.
- **Bird ↔ Airplane/Frog** — небольшой объект на однородном фоне.

Лучше всего обычно распознаются Automobile, Ship, Frog, Horse, Truck — у них характерные формы/текстуры.

## 8. Объяснение слоёв и приёмов

**Conv2d (3×3, padding=1).** Обучаемые фильтры, скользящие по изображению и ищущие локальные паттерны (границы → текстуры → части объектов → объекты). Веса общие для всех позиций, поэтому параметров гораздо меньше, чем в Linear. `padding=1` сохраняет размер карты. Два подряд 3×3 дают рецептивное поле 5×5, но с меньшим числом параметров и дополнительной нелинейностью, чем один 5×5.

**Рост числа фильтров 64 → 128 → 256.** Пространственный размер уменьшается, зато растёт «ширина»: на глубоких слоях признаков нужно больше (более абстрактные и разнообразные).

**BatchNorm2d.** Нормализует активации по батчу и канальным образом. Стабилизирует и ускоряет обучение, позволяет использовать большой learning rate (0.1), немного регуляризует. Поэтому у Conv стоит `bias=False`: BN сам добавляет обучаемый сдвиг.

**ReLU.** Нелинейность: без неё стопка свёрток — одна линейная операция. ReLU дёшев и меньше страдает от vanishing gradients, чем sigmoid/tanh.

**MaxPool2d(2).** Уменьшает карту вдвое, снижает объём вычислений, даёт небольшую инвариантность к сдвигам и расширяет рецептивное поле.

**Dropout (0.1 → 0.2 → 0.3 → 0.4).** Случайно отключает нейроны на обучении, не давая сети полагаться на отдельные признаки — борьба с переобучением. В свёрточных блоках вероятность маленькая (там меньше параметров), в полносвязной части — больше (там их больше всего).

**Flatten + Linear(4096→256) + Linear(256→10).** Flatten превращает карты признаков в вектор; первый Linear комбинирует признаки, последний даёт 10 логитов — по одному на класс.

**Нет Softmax в конце.** `CrossEntropyLoss` включает log-softmax внутри; softmax применяем вручную только при выводе уверенности.

**Data Augmentation (RandomCrop + Flip).** Искусственно увеличивает обучающую выборку, учит инвариантности к сдвигам и отражению; один из самых эффективных приёмов против переобучения на CIFAR-10.

**SGD + Nesterov momentum + weight decay.** Momentum сглаживает шумные градиенты и ускоряет движение; weight decay штрафует большие веса (L2-регуляризация).

**OneCycleLR.** Разогрев lr → пик → плавное затухание: быстрая сходимость в начале и точная настройка в конце.

**Validation и выбор лучшей эпохи.** Гиперпараметры и чекпоинт выбираем по val, чтобы test оставался честной оценкой.

### Идеи для улучшения (если не добрали до нужной accuracy)
- Больше эпох (50–100) или шире сеть (96-192-384 фильтров).
- Более сильная аугментация: Cutout (`transforms.RandomErasing`), ColorJitter.
- Label smoothing (`CrossEntropyLoss(label_smoothing=0.1)`) — снижает переуверенность.
- Ансамбль/усреднение предсказаний с отражённым вариантом картинки (TTA).
